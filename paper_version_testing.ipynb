```julia
using Distributed
procs_to_use = 15
if nprocs() <= procs_to_use
    addprocs(abs(procs_to_use-nprocs()+1))
end
;
include("const_red_MOT.jl")
using Serialization
```

```julia
prob.p.add_spontaneous_decay_kick = true
set_initial_beam_optionally_uniform!(prob, (1e-3/sqrt(2), 1e-3/sqrt(2), 1e-3), (-3e-2/sqrt(2),-3e-2/sqrt(2),0), (10/sqrt(2), 10/sqrt(2), 0), (0.1/sqrt(2), 0.1/sqrt(2), 0), [false, false, false, false, false, false])
n_trajectories = 1
detuning = -10 #MHz
detuning_repump = 0
P = 40e-3 #W
P_repump = 40e-3 #W
beam_radius = 0.825e-2 #m
Bz_gradient = -14 #G/cm
s_ratios = [3,1,3,5]
s_ratios_repump = [3,1,3,5]
pol_config = 1
pol_config_repump = 1
update_red_MOT_with_repump_params!(prob, detuning, detuning_repump, P, P_repump, beam_radius, Bz_gradient, s_ratios, s_ratios_repump, pol_config, pol_config_repump; iris_factor=0.25)

sols = distributed_solve(n_trajectories, prob, prob_func!, scan_nothing, [0])
data_quarter_iris = get_data(sols[1])
;
```

```julia
plot((data_quarter_iris.xs+data_quarter_iris.ys)*1e2/sqrt(2))
```

```julia
plot(data_quarter_iris.zs*1e2)
```
