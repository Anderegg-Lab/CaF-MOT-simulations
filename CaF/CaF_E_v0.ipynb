```julia
using QuantumStates, UnitsToValue
```

```julia
H_operator = :(
    Te * DiagonalOperator +
    BE * Rotation + 
    DE * RotationDistortion + 
    γE * SpinRotation
    # +bFX * Hyperfine_IS + 
    # cX * (Hyperfine_Dipolar/3)
);
```

```julia
parameters = @params begin
    Te = 34171.218* c * 1e2
    BE = 0.36326*c*1e2
    DE= 4.27e-7*c*1e2
    γE = 0.0557*c*1e2
    # bX = 109.1839 * 1e6
    # cX = 40.1190 * 1e6
end
# parameters.param_dict[:bFX] = parameters.bX + parameters.cX / 3;
```

```julia
QN_bounds = (label = "E", S = 1/2, I = 1/2, Λ = 0, N = 0:4)
basis = order_basis_by_m(enumerate_states(HundsCaseB_LinearMolecule, QN_bounds))
;
```

```julia
CaF_E_v0 = @time Hamiltonian(basis=basis, operator=H_operator, parameters=parameters)
@time full_evaluate!(CaF_E_v0)
@time solve!(CaF_E_v0)
;
```

```julia
@time save_to_file(CaF_E_v0, "CaF_E_v0", "")
```
