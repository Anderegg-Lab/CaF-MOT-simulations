```julia
using QuantumStates, UnitsToValue
```

```julia
H_operator = :(
    Te * DiagonalOperator +
    BD * Rotation + 
    DD * RotationDistortion + 
    γD * SpinRotation
    #+ bFX * Hyperfine_IS + 
    # cX * (Hyperfine_Dipolar/3)
);
```

```julia
parameters = @params begin
    Te = 30158.617 * c * 1e2
    BD = 0.36544*c*1e2
    DD = 4.5e-7*c*1e2
    γD = -0.00899*c*1e2
    # bX = 109.1839 * 1e6
    # cX = 40.1190 * 1e6
end
# parameters.param_dict[:bFX] = parameters.bX + parameters.cX / 3;
```

```julia
QN_bounds = (label = "D", S = 1/2, I = 1/2, Λ = 0, N = 0:4)
basis = order_basis_by_m(enumerate_states(HundsCaseB_LinearMolecule, QN_bounds))
;
```

```julia
CaF_D_v0 = @time Hamiltonian(basis=basis, operator=H_operator, parameters=parameters)
@time full_evaluate!(CaF_D_v0)
@time solve!(CaF_D_v0)
;
```

```julia
@time save_to_file(CaF_D_v0, "CaF_D_v0", "")
```
