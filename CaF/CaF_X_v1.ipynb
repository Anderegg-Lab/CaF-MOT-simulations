```julia
using QuantumStates, UnitsToValue
```

```julia
H_operator = :(
    TeX * DiagonalOperator +
    BX * Rotation + 
    DX * RotationDistortion + 
    γX * SpinRotation +
    bFX * Hyperfine_IS + 
    cX * (Hyperfine_Dipolar/3)
);
```

```julia
parameters = @params begin
    TeX = 582.8478*1e2*c # converted to Hz, value in cm^-1 from https://doi.org/10.1006/jmsp.1999.7909 
    BX = 0.34005359*1e2*c # converted to Hz, value in cm^-1 from https://doi.org/10.1006/jmsp.1999.7909 
    DX = 0.014060 * 1e6 #change from v=0 is small
    γX = 39.65891 * 1e6 #change from v=0 is small
    bX = (109.1839-1.41521) * 1e6 # from Childs, Goodman, Goodman, 1981
    cX = (40.1190+1.0562) * 1e6 # from Childs, Goodman, Goodman, 1981
end
;
parameters.param_dict[:bFX] = parameters.bX + parameters.cX / 3;
```

```julia
QN_bounds = (label = "X", v_1 = 1, S = 1/2, I = 1/2, Λ = 0, N = 0:3)
basis = order_basis_by_m(enumerate_states(HundsCaseB_LinearMolecule, QN_bounds))
;
```

```julia
CaF_X_v1 = @time Hamiltonian(basis=basis, operator=H_operator, parameters=parameters)
@time full_evaluate!(CaF_X_v1)
@time solve!(CaF_X_v1)
;
```

```julia
@time save_to_file(CaF_X_v1, "CaF_X_v1", "")
```
