```julia
using QuantumStates, UnitsToValue
```

```julia
H_operator = :(
    Te * DiagonalOperator +
    BX * Rotation + 
    DX * RotationDistortion + 
    γX * SpinRotation +
    bFX * Hyperfine_IS + 
    cX * (Hyperfine_Dipolar/3)
);
```

```julia
# parameters = @params begin
#     Te = 18844.5 * c * 1e2
#     BX = 10303.988 * 1e6
#     DX = 0.014060 * 1e6
#     γX = 39.65891 * 1e6
#     bX = 109.1839 * 1e6
#     cX = 40.1190 * 1e6
# end
# parameters.param_dict[:bFX] = parameters.bX + parameters.cX / 3;
parameters = @params begin
    Te = 18841.3 * c * 1e2
    BX = 10263.2 * 1e6
    DX = 0.01444 * 1e6
    γX = -1.64*c*1e2 # from https://doi-org.libproxy1.usc.edu/10.1006/jmsp.1999.7909
    bX = 20 * 1e6 # from https://arxiv.org/pdf/1509.04021
    cX = 50 * 1e6 # from https://arxiv.org/pdf/1509.04021
end
parameters.param_dict[:bFX] = parameters.bX + parameters.cX / 3;
```

```julia
QN_bounds = (label = "B", S = 1/2, I = 1/2, Λ = 0, N = 0:4)
basis = order_basis_by_m(enumerate_states(HundsCaseB_LinearMolecule, QN_bounds))
;
```

```julia
CaF_B_v0 = @time Hamiltonian(basis=basis, operator=H_operator, parameters=parameters)
@time full_evaluate!(CaF_B_v0)
@time solve!(CaF_B_v0)
;
```

```julia
@time save_to_file(CaF_B_v0, "CaF_B_v0", "")
```
